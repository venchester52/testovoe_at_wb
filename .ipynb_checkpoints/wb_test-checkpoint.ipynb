```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data_.csv')
df.head()
```

```python
print('Информация о датафрейме:')
df.info()
```

```python
print("\nКоличество пропусков по столбцам:")
df.isnull().sum()
```

```python
# Визуализация выбросов
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.boxplot(y=df['age'], ax=axes[0]).set_title('Возраст (age)')
sns.boxplot(y=df['income'], ax=axes[1]).set_title('Доход (income)')
sns.boxplot(y=df['score_bki'], ax=axes[2]).set_title('Скор БКИ (score_bki)')
plt.show()

# Функция для подсчета границ выбросов по IQR
def get_outliers_bounds(column):
    q1 = column.quantile(0.25)
    q3 = column.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    return lower_bound, upper_bound

# Очистка от жестких выбросов по доходу (как самому подверженному выбросам признаку)
lower_income, upper_income = get_outliers_bounds(df['income'])
df_clean = df[(df['income'] >= lower_income) & (df['income'] <= upper_income)].copy()

print(f"Размер исходных данных: {df.shape}")
print(f"Размер данных после очистки выбросов дохода: {df_clean.shape}")
```

```python
# Выбираем числовые признаки
num_cols = ['age', 'decline_app_cnt', 'bki_request_cnt', 'income', 'score_bki', 'region_rating']

plt.figure(figsize=(10, 8))
sns.heatmap(df_clean[num_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Матрица корреляций числовых признаков')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df_clean, x='age', hue='default', common_norm=False, fill=True)
plt.title('Распределение возраста в зависимости от дефолта')
plt.show()

print("Средний возраст по группам:")
display(df_clean.groupby('default')['age'].mean())
```
